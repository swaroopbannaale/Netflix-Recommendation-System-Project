```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# importing and reading data file
df = pd.read_csv('/content/drive/MyDrive/Netflix dataset/combined_data_1.txt.zip',header = None, names = ['Cust_Id', 'Rating'], usecols = [0,1])

# converting rating column to a float
df['Rating'] = df['Rating'].astype(float)
```

```python
df
```

```python
# datatype of columns
df.dtypes
```

```python
# Checking all the movie id's present in our data
movie_count = df[df['Cust_Id'].str.contains(':')].count()[0]
```

```python
movie_count
```

```python
# shape of data
df.shape
```

```python
df.head()
```

```python
# no of unique customer id in our data
cust_count=df['Cust_Id'].nunique()
cust_count
```

```python
# total cust_count
# In this total customer id count the movid id will be also counted.
cust_count = cust_count-movie_count
cust_count
```

```python
# total rating
rating_count = df['Rating'].count()
rating_count
```

```python
# Segregation of our data
movie_np = []
for x in df['Cust_Id']:
    if ':' in x:
        y = x.replace(':',"")
    movie_np.append(y)
```

```python
df['MovieID'] = movie_np
```

```python
df
```

```python
df.dropna(inplace=True)
df
```

```python
df['CustID']  = df['Cust_Id'].astype(int)
df['MovieID']  = df['MovieID'].astype(int)
```

```python
df.info()
```

```python
# filteration
data_cust_summary = df.groupby('CustID')['Rating'].count()
data_cust_summary
```

```python
# customers with more than 36 movie views
movie_benchmark = round(data_cust_summary.quantile(0.60))
movie_benchmark
```

```python
Rejected_cust = data_cust_summary[data_cust_summary<movie_benchmark].index
# When we use groupby fxn the column becomes index
Rejected_cust
```

```python
# Filtering our movie data
# First we will get the count of ratings for every movie.
data_movie_summary = df.groupby('MovieID')['Rating'].count()
data_movie_summary
```

```python
cust_benchmark = round(data_movie_summary.quantile(0.60))
cust_benchmark
```

```python
Rejected_movies = data_movie_summary[data_movie_summary<cust_benchmark].index
Rejected_movies
```

```python
# removing rejected data and upload as original
df = df[~df['MovieID'].isin(Rejected_movies)]
df
```

```python
df = df[~df['CustID'].isin(Rejected_cust)]
df
```

```python
df = df.drop(columns=['CustID'])
df
```
